Q1 : What is the difference between multithreading and multiprocessing?
- Multithreading means executing multiple threads within a single process, where all threads share the same memory space.
 On the other hand, multiprocessing involves running multiple processes simultaneously, and each process has its own separate memory. In multithreading, threads share resources, making it memory efficient but also risky—if one thread crashes, it can affect the whole process. In contrast, in multiprocessing, each process runs independently, so the failure of one process does not impact others. Multithreading is generally faster for I/O-bound tasks such as reading files or making network requests, while multiprocessing is more suitable for CPU-bound tasks that require heavy computation. In Python, multithreading is implemented using the threading module, and multiprocessing is done using the multiprocessing module.






Q2: What are the challenges associated with memory management in Python?
- Memory management in Python comes with several challenges.
1. Python uses automatic garbage collection, but it may not immediately free unused memory, leading to higher memory usage.
2. Reference cycles, where objects reference each other, can make it difficult for the garbage collector to release memory.
3. Large data structures like lists or dictionaries can consume a lot of memory if not handled properly.
4. Memory fragmentation can occur because Python objects of different sizes are stored in various memory blocks.
5. Managing global and local variables efficiently is also challenging since unused variables may stay in memory longer than needed.
6. Programmers must be careful with object references and data copies to avoid memory leaks and performance issues.

Q6: What are the differences between NumPy arrays and Python lists?
- NumPy arrays and Python lists are both used to store collections of data.
- NumPy arrays are homogeneous, which means all elements must have the same data type.
-  Python lists, however, can store different types of elements in the same list.
-  NumPy arrays are faster and more memory-efficient because they store data in a continuous block of memory, while lists are slower for large-scale numerical operations.
- NumPy supports vectorized operations, so arithmetic can be done on all elements without loops.
-  Lists require loops or comprehensions to perform similar operations. NumPy also provides many built-in functions for math, statistics, and linear algebra, which are not directly available with Python lists.

Q7:Explain the difference between apply() and map() in Pandas.
- In Pandas, both apply() and map() are used to apply functions to data, but they are used differently.

- map() is mainly used with Series and is applied element-wise. It can take a function, dictionary, or Series to transform each element individually.

- apply() can be used with both Series and DataFrame. For a Series, it works like map() and applies a function to each element. For a DataFrame, it can apply a function row-wise or column-wise, giving more flexibility.

Main difference : map() is simpler and limited to element-wise operations on a Series, while apply() is more versatile and can work on both Series and DataFrames with custom functions.

```python
#Q10: Calculate the correlation matrix using Seaborn and visualize it with heatmap

# Import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Math": [90, 80, 70, 85, 95],
    "Physics": [85, 78, 72, 88, 94],
    "Chemistry": [88, 82, 75, 90, 92],
    "Biology": [92, 79, 68, 84, 91]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Calculate correlation matrix
corr_matrix = df.corr()

# Display correlation matrix
print("Correlation Matrix:\n", corr_matrix)

# Visualize with heatmap
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
